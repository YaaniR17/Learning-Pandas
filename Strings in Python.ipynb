### Making our Strings either UPPERCASE or LOWERCASE.

The first string methods we will meet are:

* .upper()
* .lower()

As it says on the tin....it will return a supplied string either all uppercase or all lowercase.

An example:

---

```python
test_value_lower = "i am a lower string"
test_value_upper = "I AM AN UPPER STRING"

print(test_value_lower.upper()) # Convert test_value_lower to uppercase
print(test_value_upper.lower()) # Convert test_value_upper to lowercase
```

---

As we can see...each printed string is the opposite case of what it was when it started.

This can be useful with user input. It's a lot simpler to deal with either all uppercase or lowercase.

### Is the string already upper or lowercase?

Out next methods we are going to meet are:

* .isupper()
* .islower()

Each method will return true if all the charecters in the string are, respectively, all upper or all lower case. 

Only letter are checked though. The methods will ignore non letter charecters.

An example:

---

```python
test_value_lower = "i am a lower string"
test_value_upper = "I AM AN UPPER STRING"
test_value_neither = "I Am Neither Up or Down"


print(test_value_lower.isupper()) # Check case with iflower()
print(test_value_upper.islower()) # Check case with ifupper()
print("\n")
print(test_value_lower.islower()) # Check case with iflower()
print(test_value_upper.isupper()) # Check case with ifupper()
print("\n")
print(test_value_neither.islower()) # Check case with iflower()
print(test_value_neither.isupper()) # Check case with ifupper()
```

### But what type of string are we dealing with?

Out next methods to meet are:

* .isalpha()
* .isalnum()
* .isdigit()
* .isnumeric()


_.isalpha()_ returns true if the contents of the string are only alphabet letters (a-z).


_.isalnum()_ returns true if the contents of the string are alphanumeric (a-z 0-9).

_.isdigit()_ returns true if the contents of the string are standard digits (0-9).

_.isnumeric()_ Similar to _isdigit()_ but returns true on a wider formatting of digits such as fractions.

An example:

---

```python
# An example of .isalpha() and .isalnum()

test_value = "College"
test_value2 = "College 2025"

print(test_value.isalpha())
print(test_value2.isalpha())
print("\n")
print(test_value.isalnum())
print(test_value2.isalnum())
print("\n")
# An example of is .digit() and .isnumeric()

test_value = "1234"
test_value2 = "1234.5"

print(test_value.isdigit())
print(test_value2.isnumeric())
print("\n")
print(test_value.isnumeric())
print(test_value2.isdigit())
```

### I'm not sure where a bit of string is!

We can use:

* _.find(value, start, end)_
* _.index(value, start, end)_

to search strings for a __substring__ and retreive it's location.

Both are essentially the same in that they search for a _value_ from an optional start and endpoint and return the index of the start of the _substring__.

Where they differ is that _.find()_ will return a _-1_ error if the __substring__ is not found and _.index()_ will raise and exception which will cause your code to crash unless managed in a _try...except_ structure.

An example:

---

```python
test_value = "A cat sat on the mat"

# example of .find()

print(test_value.find("sat"))
print(test_value.find("dog"))
print("\n")

# example of index
# (using a try...except)

print(test_value.index("sat"))
try:
    print(test_value.index("dog"))
except Exception as e: print(e)
print("\n")

# example of .index() without a try...except
print("should have an Error now...this is normally not good :) ")
print(test_value.index("dog"))
```

### Strings must be changed...change all the strings!

Changing the contents of strings is simple enough...let's have a look at the following methods:

* _.split(separator, maxsplit)_

* _.replace(oldvalue, newvalue, count)_

* _.strip(characters)_

_.split()_ splits a string into a list. We need to provide a *seperator* and, optionally, also how many splits to do.

_.replace()_ replaces a substring with a specified substring. An optional count will tell it how many occurances to change. Default it to all substrings it finds.

_.strip()_ removes the leading and trailing whitespaces from a string. It can remove other charecters if told to do so.

---

```python
# example of a .split()
test_value = "A cat sat on the mat"

print(type(test_value))
print(test_value)

new_test_value = test_value.split(" ")

print(type(new_test_value))
print(new_test_value)

# example of a .replace()
print("\n")
print(test_value)
print(test_value.replace("cat","dog"))

## example of a .strip()
new_test_value = "          A cat sat on the mat          "

print("\n")
# remove the leading "A"
print(test_value.strip("A"))
# remove the trailing "t"
print(test_value.strip("t"))
# remove all leading and trailing spaces
print("\n")
print(new_test_value)
print(new_test_value.strip())


```

### Glueing Strings together

We can also join strings together in two seperate ways:

* _.format()_
* Concatenation using the + operator 

Out of the two, concatenation is the easier way but _.format()_ is a lot more powerful and is the more accepted way.

It uses the concepts of placeholders in your strings that Python will automatically fill in with the correct values.

_.format()_ will also handle the automatic casting of datatypes to become strings where as, with concatenation you have to manually manage such casting (coversions).

Let's have a look:

---

```python
test_value = "A cat sat on the mat"
second_test_value = "and it was most comfortable"

# example of concatenation
print(test_value+second_test_value)

# We have an issue above in that there is no space inbetween the to strings!
print(test_value+" "+second_test_value)
print("\n")
# example of .format()

output = "A {animal} sat on the {furniture} and it was most comfortable"
print(output)
print("\n")
print(output.format(animal = "mouse", furniture = "windowsill"))
print("\n")
print(output.format(animal = "crocodile", furniture = "log"))
print("\n")
print("A {person} {did_this} today!".format(person = "student", did_this = "learned something"))

```

---

As we can see, using a string formatting with _.format()_ is a bit more flexible in use. We can potentially store strings ready for use but update them just before they are displayed.
